```python
user_anime=anime.pivot(index='anime_id',columns='genre',values='rating')
user_anime
```

```python
user_anime.fillna(0,inplace=True)
```

```python
user_anime
```

```python
from sklearn.metrics import pairwise_distances
from scipy.spatial.distance import cosine,correlation
```

```python
user_sim=1-pairwise_distances(user_anime.values,metric='cosine')
user_sim
```

```python
np.fill_diagonal(user_sim,0)
```

```python
user_sim
```

```python
user_sim_df=pd.DataFrame(user_sim)
user_sim_df
```

```python
user_sim_df.index=anime.anime_id.unique()
user_sim_df.columns=anime.anime_id.unique()
```
