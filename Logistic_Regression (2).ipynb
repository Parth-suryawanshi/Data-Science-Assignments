```python
df1['Sex']=df1['Sex'].map({'male':0,'female':1})
```

```python
df1=pd.get_dummies(df1,columns=['Embarked'])
```

```python
df1.dtypes
```

```python
# Model Building
```

```python
df1.head()
```

```python
X = df1.drop('Name',axis=1)

Y = df1['Survived']
```

```python
X=X.drop('Survived',axis=1)
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_val, y_train, y_val = train_test_split( X, Y, test_size=0.2, random_state=42, stratify=Y)
```

```python
classifier=LogisticRegression()
classifier.fit(X_train,y_train)
```

```python
# Model Evaluation
```

```python
# Predict for X dataset
```

```python
y_pred=classifier.predict(X)
```

```python
y_pred
```

```python
y_pred_df=pd.DataFrame({'Actual':Y,'Predicted':y_pred})
y_pred_df
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
accuracy_score=accuracy_score(Y,y_pred)
precision_score=precision_score(Y,y_pred)
recall_score=recall_score(Y,y_pred)
f1_score=f1_score(Y,y_pred)
roc_auc_score=roc_auc_score(Y,y_pred)
print(f"Accuracy Score: {accuracy_score}")
print(f"Precision Score: {precision_score}")
print(f"Recall Score: {recall_score}")
print(f"F1 Score: {f1_score}")
print(f'ROC AUC Score: {roc_auc_score}')
```

```python
from sklearn.metrics import roc_curve
import numpy as np
```

```python
fpr,tpr,thresholds=roc_curve(Y,classifier.predict_proba(X)[:,1])
plt.plot(fpr,tpr,color='red',label='ROC')
plt.plot(np.linspace(0,1,100),np.linspace(0,1,100),label='baseline')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC_Curve')
plt.legend()
plt.show()



```

```python
# Coefficients
coefficients = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': classifier.coef_[0],
    'Odds Ratio': np.exp(classifier.coef_[0])
})

print(coefficients)
```

```python
# positive coefficient → increases survival probability.
#Negative coefficient → decreases survival probability.
```

```python
'''
Higher absolute value → more influence

Positive sign → increases survival probability

Negative sign → decreases survival probability
```
