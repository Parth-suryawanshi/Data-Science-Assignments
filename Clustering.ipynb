```python
# Outlier_detection
def Outlier_detection(data,colname):
  q1=data[colname].quantile(0.25)
  q3=data[colname].quantile(0.75)
  iqr=q3-q1
  upper_extreme=q3+(1.5*iqr)
  lower_extreme=q1-(1.5*iqr)
  return upper_extreme,lower_extreme
```

```python
Outlier_detection(df,'Balance')
```

```python
# kMeans Clustering

from sklearn.cluster import KMeans
```

```python
# standardisation for z_score calculation
Airline=pd.DataFrame(sc.fit_transform(df),columns=df.columns)
Airline
```

```python
wcss=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',max_iter=500,random_state=0)
  kmeans.fit(Airline)
  wcss.append(kmeans.inertia_)


plt.plot(range(1,11),wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('wcss')
plt.show()
```
