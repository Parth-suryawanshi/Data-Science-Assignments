```python
import numpy as np
from scipy import stats

# Data provided
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])

# Calculate sample statistics
n = len(data)
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1) # ddof=1 for sample standard deviation

print(f"Sample Size (n): {n}")
print(f"Sample Mean: {sample_mean:.3f} million characters")
print(f"Sample Standard Deviation: {sample_std:.3f} million characters")
print("\n" + "-"*50 + "\n")

# --- Assignment Task a: Build 99% Confidence Interval Using Sample Standard Deviation ---
print("Assignment Task a: 99% Confidence Interval Using Sample Standard Deviation")

# Rationale for using t-distribution:
# We use the t-distribution because the sample size (n=15) is small (typically n < 30)
# and the population standard deviation is unknown. The t-distribution accounts for the
# increased uncertainty due to estimating the population standard deviation from the sample.

# Steps:
# 1. Determine degrees of freedom (df = n - 1).
# 2. Find the t-critical value for a 99% confidence level with the calculated degrees of freedom.
# 3. Calculate the standard error of the mean (SE = sample_std / sqrt(n)).
# 4. Calculate the margin of error (ME = t_critical * SE).
# 5. Construct the confidence interval (sample_mean \u00b1 ME).

df = n - 1 # Degrees of freedom
confidence_level = 0.99
alpha = 1 - confidence_level # Significance level

# Find the t-critical value (two-tailed)
t_critical = stats.t.ppf(1 - alpha/2, df)

# Calculate Standard Error of the Mean
standard_error_t = sample_std / np.sqrt(n)

# Calculate Margin of Error
margin_of_error_t = t_critical * standard_error_t

# Construct the Confidence Interval
confidence_interval_t_lower = sample_mean - margin_of_error_t
confidence_interval_t_upper = sample_mean + margin_of_error_t

print(f"Degrees of Freedom (df): {df}")
print(f"t-critical value for 99% CI: {t_critical:.3f}")
print(f"Standard Error (t-dist): {standard_error_t:.3f}")
print(f"Margin of Error (t-dist): {margin_of_error_t:.3f}")
print(f"99% Confidence Interval (t-dist): ({confidence_interval_t_lower:.3f}, {confidence_interval_t_upper:.3f}) million characters")

print("\n" + "-"*50 + "\n")

# --- Assignment Task b: Build 99% Confidence Interval Using Known Population Standard Deviation ---
print("Assignment Task b: 99% Confidence Interval Using Known Population Standard Deviation")

# Given population standard deviation
population_std = 0.2

# Rationale for using z-distribution:
# We use the z-distribution (standard normal distribution) because the population
# standard deviation is known. Even with a small sample size, when the population
# standard deviation is known, the sampling distribution of the mean follows a
# normal distribution (or can be approximated by it).

# Steps:
# 1. Find the z-critical value for a 99% confidence level.
# 2. Calculate the standard error of the mean (SE = population_std / sqrt(n)).
# 3. Calculate the margin of error (ME = z_critical * SE).
# 4. Construct the confidence interval (sample_mean \u00b1 ME).

# Find the z-critical value (two-tailed)
z_critical = stats.norm.ppf(1 - alpha/2)

# Calculate Standard Error of the Mean
standard_error_z = population_std / np.sqrt(n)

# Calculate Margin of Error
margin_of_error_z = z_critical * standard_error_z

# Construct the Confidence Interval
confidence_interval_z_lower = sample_mean - margin_of_error_z
confidence_interval_z_upper = sample_mean + margin_of_error_z

print(f"Known Population Standard Deviation: {population_std:.3f} million characters")
print(f"z-critical value for 99% CI: {z_critical:.3f}")
print(f"Standard Error (z-dist): {standard_error_z:.3f}")
print(f"Margin of Error (z-dist): {margin_of_error_z:.3f}")
print(f"99% Confidence Interval (z-dist): ({confidence_interval_z_lower:.3f}, {confidence_interval_z_upper:.3f}) million characters")
```
